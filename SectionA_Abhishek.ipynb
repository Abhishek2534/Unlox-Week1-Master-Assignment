Q1. What is the difference between the is operator and the == operator? Give one example where two variables return True for == but False for is.
Answer:
The == operator checks whether two objects have equal values, whereas the is operator checks whether they refer to the same object in memory.



```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)  # True
print(a is b)  # False
print(a is c)  # True
```

Q5. Why is dictionary lookup O(1) on average while searching for a value in a list using in is O(n)? Explain in plain English without using the word "hash" – then explain it again WITH the word "hash".

Answer:
A dictionary can quickly locate an item using its key, while a list may need to check each element one by one until it finds the required value.

Using the word hash:

A dictionary uses a hash function to convert a key into a location in its internal structure. This makes average lookup time O(1), whereas searching a list takes O(n) because it may check every element.

Q7. What is the difference between list.append(x), list.extend(x), and list.insert(i, x)? For each, show what happens when x = [4, 5] is added to a = [1, 2, 3].

Answer:

1. append():adds the entire list as a single element at the end.

```python
a = [1, 2, 3]
x = [4, 5]
a.append(x)
print(a)
```

2. extend() adds each element separately.

```python
a = [1, 2, 3]
x = [4, 5]
a.extend(x)
print(a)
```

3. insert() adds an element at a specified position.

```python
a = [1, 2, 3]
x = [4, 5]
a.insert(1, x)
print(a)
```

Q10. Why is NumPy significantly faster than Python lists for numerical operations? Give 3 reasons, at least one should mention memory layout.

Answer:

Memory layout: NumPy stores elements in compact, usually contiguous memory, reducing memory overhead.

Vectorization: NumPy performs operations on entire arrays without requiring Python-level loops for each element.

Optimized implementation: NumPy uses optimized compiled code for numerical operations, making calculations faster.
